```python
#Practical 4a -Travelling salesman(greedy approach)
from sys import maxsize
from itertools import permutations

V = 4

# Function to solve the Traveling Salesman Problem
def travellingSalesmanProblem(graph, s):

    # Store all vertices apart from the source vertex
    vertex = [i for i in range(V) if i != s]

    # Store minimum path cost
    min_path = maxsize

    # Generate all permutations of the vertices
    for i in permutations(vertex):

        # Store current path weight (cost)
        current_pathweight = 0

        # Compute current path weight
        k = s
        for j in i:
            current_pathweight += graph[k][j]
            k = j

        # Return to the starting city
        current_pathweight += graph[k][s]

        # Update minimum path cost
        min_path = min(min_path, current_pathweight)

    return min_path


# Driver Code
if __name__ == "__main__":

    # Matrix representation of the graph
    graph = [
        [0, 10, 15, 20],
        [10, 0, 35, 25],
        [15, 35, 0, 30],
        [20, 25, 30, 0]
    ]

    s = 0

    print("Minimum Travelling Cost:", travellingSalesmanProblem(graph, s))
```
